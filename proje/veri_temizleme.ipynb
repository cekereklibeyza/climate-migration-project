```python
df_asya.describe()
```

```python
# eksik değer ve duplicate kontrolü
print(df_asya.isnull().sum().sum(), "eksik hücre")
print(df_asya.duplicated(subset=["Year","Country of Asylum","Country of Origin"]).sum(), "duplicate satır")
```

Female/Male kırılım sütunlarının toplamı bazı satırlarda genel `Total` ile uyuşmuyor - UNHCR bazı yıllarda sadece toplam rakamı paylaşmış, kırılımı paylaşmamış. Bu yüzden kırılım sütunlarını kullanmayacağız, sadece `Total`'e güveneceğiz.

```python
tutarsiz = df_asya[df_asya["Female Total"] + df_asya["Male Total"] != df_asya["Total"]]
print(len(tutarsiz), "satırda kırılım toplamı Total ile uyuşmuyor")
```

`Total` sütununda IQR ile aykırı değer kontrolü - bunlar genelde gerçek kriz/savaş olayları olduğu için silinmiyor, sadece not ediliyor.

```python
Q1, Q3 = df_asya["Total"].quantile([0.25, 0.75])
IQR = Q3 - Q1
ust = Q3 + 1.5*IQR
print(((df_asya["Total"] > ust).sum()), "satır aykırı değer sınırının üstünde")
df_asya.nlargest(5, "Total")[["Year","Country of Asylum","Country of Origin","Total"]]
```

```python
sutunlar_atilacak = [
    "Female 0 - 4", "Female 5 - 11", "Female 12 - 17", "Female 18 - 59", "Female 60+", "Female Other", "Female Total",
    "Male 0 - 4", "Male 5 - 11", "Male 12 - 17", "Male 18 - 59", "Male 60+", "Male Other", "Male Total"
]
df_asya_temiz = df_asya.drop(columns=sutunlar_atilacak)
df_asya_temiz.to_csv("goc_asya_pasifik_temiz.csv", index=False)
df_asya_temiz.shape
```

## Sıcaklık

Bu veri diğerlerinden farklı - her satır bir ülke, her sütun bir yıl (wide format). Göç verileriyle birleştirebilmek için long formata çevireceğiz.

```python
sicaklik = pd.read_excel("/content/yıllık ortalama sıcaklık.xlsx", sheet_name="all")
print(sicaklik.shape)
sicaklik.head()
```

İki satır (Bouvet Island, Saint-Barthélemy) 125 yılın tamamında boş - bunları atıyoruz.

```python
yil_sutunlari = [c for c in sicaklik.columns if c not in ["code","name"]]
bos_satirlar = sicaklik[sicaklik[yil_sutunlari].isna().all(axis=1)]
print(bos_satirlar[["code","name"]])
```

```python
sicaklik = sicaklik[~sicaklik["code"].isin(["BVT","BLM"])].reset_index(drop=True)

sicaklik_long = sicaklik.melt(id_vars=["code","name"], var_name="Year", value_name="Avg_Temp")
sicaklik_long["Year"] = sicaklik_long["Year"].str.split("-").str[0].astype(int)
sicaklik_long = sicaklik_long.rename(columns={"code":"Country ISO","name":"Country"})[["Year","Country","Country ISO","Avg_Temp"]]

print(sicaklik_long.shape)
print(sicaklik_long["Avg_Temp"].isna().sum(), "eksik değer")
```

Kalan tek eksik değer (Mongolia, 1925) - komşu yıllardan lineer enterpolasyonla dolduruyoruz.

```python
sicaklik_long = sicaklik_long.sort_values(["Country ISO","Year"])
sicaklik_long["Avg_Temp"] = sicaklik_long.groupby("Country ISO")["Avg_Temp"].transform(lambda x: x.interpolate())
sicaklik_long = sicaklik_long.sort_values(["Year","Country ISO"]).reset_index(drop=True)

print(sicaklik_long["Avg_Temp"].isna().sum(), "eksik değer kaldı")
sicaklik_long.to_csv("yillik_ortalama_sicaklik_temiz.csv", index=False)
```

## Afet (kuraklık, sel, fırtına, aşırı sıcaklık)

Bu veri olay bazlı - her satır tek bir afet. Göç verisiyle aynı kapsama (2001-2025) çekip Yıl+Ülke bazında özetleyeceğiz.

```python
afet = pd.read_excel("/content/temiz kuraklık, sel, sis, fırtına, aşırı sıcaklık verileri tüm dünya.xlsx")
print(afet.shape)
afet["Disaster Type"].value_counts()
```

Start Date, End Date'ten sonra olamaz. Bu koşulu sağlamayan satırlar var, kontrol edelim.

```python
afet["sd"] = pd.to_datetime(afet["Start Date"])
afet["ed"] = pd.to_datetime(afet["End Date"])
ters = afet[afet["sd"] > afet["ed"]]
print(len(ters), "satırda Start Date > End Date")
```

Bir kayıtta (DisNo. 1988-0424-VEN) yıl alanı bozuk - DisNo olayın 1988'e ait olduğunu gösteriyor, Start Date'i End Date ile aynı yapıyoruz. Diğer 187 satır ise Start/End'in yer değiştirmesi - swap ettiğimizde süre 1-335 gün arası çıkıyor (mantıklı), bu yüzden swap ile düzeltiyoruz.

```python
ven_mask = afet["DisNo."] == "1988-0424-VEN"
afet.loc[ven_mask, "Start Date"] = afet.loc[ven_mask, "End Date"]

afet["sd"] = pd.to_datetime(afet["Start Date"])
afet["ed"] = pd.to_datetime(afet["End Date"])
ters_mask = (afet["sd"] > afet["ed"]) & (~ven_mask)
afet.loc[ters_mask, ["Start Date","End Date"]] = afet.loc[ters_mask, ["End Date","Start Date"]].values

afet["sd"] = pd.to_datetime(afet["Start Date"])
afet["ed"] = pd.to_datetime(afet["End Date"])
print((afet["sd"] > afet["ed"]).sum(), "kalan ters tarih")

afet["Start_Year"] = afet["sd"].dt.year
afet = afet[(afet["Start_Year"]>=2001)&(afet["Start_Year"]<=2025)].drop(columns=["sd","ed"]).reset_index(drop=True)
print(afet.shape)

afet.to_csv("afet_verisi_temiz.csv", index=False)
```

Bir ülke-yılda birden fazla afet olabildiği için (örn. bazı yıllarda 30+), göç verisiyle birleştirmeden önce Yıl+Ülke bazında özetliyoruz.

```python
afet_ozet = afet.groupby(["Start_Year","ISO","Country"]).agg(
    Afet_Sayisi=("DisNo.","count"),
    Kuraklik_Sayisi=("Disaster Type", lambda x: (x=="Drought").sum()),
    Sel_Sayisi=("Disaster Type", lambda x: (x=="Flood").sum()),
    Firtina_Sayisi=("Disaster Type", lambda x: (x=="Storm").sum()),
    Asiri_Sicaklik_Sayisi=("Disaster Type", lambda x: (x=="Extreme temperature").sum()),
    Toplam_Olum=("Total Deaths","sum"),
    Toplam_Etkilenen=("Total Affected","sum"),
    Toplam_Hasar_1000USD=("Total Damage ('000 US$)","sum"),
).reset_index().rename(columns={"Start_Year":"Year"})

print(afet_ozet.shape)
print(afet_ozet["Afet_Sayisi"].sum() == len(afet))

afet_ozet.to_csv("afet_yil_ulke_ozet.csv", index=False)
```
